```python
with open ('example.txt', 'r') as file:
    
    for line in file:
        print(line.strip())
```

```python
## writing in a file (overwrite)
with open ('example.txt', 'w') as file:
    file.write('hellow world\n')
    file.write('this is a new line')
```

```python
## writing without overwrite 
with open ('example.txt','a') as file:
    file.write('apend function taking place')
```

```python
##writing multiple lines
line=['first  line \n','second line  \n']

with open('example.txt', 'a') as file:
    file.writelines(line)
```

```python
## binary files 

data = b'\x01\x02\x10'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
##reading source file content and write in destination file  
with open ('example.txt', 'r') as source_file:
    content = source_file.read()
    
with open ('destination.txt', 'w') as des_file:
    des_file.write(content)
    
```

```python
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
        return line_count, word_count,char_count
    
file_path = 'example.txt'
lines , words , characters = count_text_file('example.txt')
print(f'lines: {lines}, words: {words}, char: {characters}')
```
